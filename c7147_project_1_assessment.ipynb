## 🧬 Load reference and reads data into variables
Below, we provide some helper code to load the reference from `reference.txt` and the reads from `reads.csv`.

❗ **Do not change the function’s behavior** (you may read it, of course!).

The code creates two variables from the loaded files:
- **reference**: a string containing the 200-nt reference sequence
- **reads**: a list of dictionaries containing the 6,000 reads

````python
##### Constants

REF_PATH = "reference.txt"
READS_PATH = "reads.csv"

##### Functions

def load_reference(path: str) -> str:
    """Load reference nucleotide sequence from file at path to reference string."""
    with open(path) as f:
        lines = [ln.strip() for ln in f if ln.strip()]
    if lines[0].startswith('>'):
        lines = lines[1:]
    ref = "".join(lines).upper()
    if len(ref) != 200:
        raise ValueError(f"Reference must be 200 nt, got {len(ref)}")
    return ref


def load_reads(path: str) -> list:
    """Load sample NGS reads from file at path to reads list."""
    with open(path, newline="") as f:
        reader = csv.DictReader(f)
        reads = list(reader)
    return reads

##### Main code

# Load reference and reads
reference = load_reference(REF_PATH)
reads = load_reads(READS_PATH)

# Report data structures
print("Reference length:", len(reference))
print("Total no. of reads:", len(reads))
print("Keys in each read dictionary:", reads[0].keys())
print("Samples loaded:", sorted({read["sample"] for read in reads}))

# Print sample sequences
print("\nSample read data:")
for row in reads[:5]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
print("...")
for row in reads[-5:]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
````

## Step 1 — Validate each read and record any errors
## Function: `validate_read(sequence, reference)`

### 👍 You can assume:
- `reference` is a non-empty string containing only uppercase A/C/G/T.
- `sequence` is a string, but nothing about its length or contents.

### 🎯 Aims:
- Check that an individual read is the same length as the reference and contains only valid characters (A/C/G/T)
- "Repair" any reads that contain valid lower-case characters -> upper-case
- Return a dictionary with the following keys:

    - `sequence`: the upper-case version of the sequence input, including when invalid.
    - `valid`: `True` only when the length matches the reference and every character is A/C/G/T.
    - `errors`: a list of strings. Include `"wrong_length"` first if lengths differ, then `"invalid_characters"` if any character is outside A/C/G/T. Include both when both apply. A valid read has `[]`.

### 🔎 Expected outputs

Examples for reference `"ATGC"`:
| Input (read) | Input type | Output |
|---|---|:---|
| `"ATGC"` | Correct | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `"atgc"` | Lowercase | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `""` | Empty read | `{"sequence": "", "valid": False, "errors": ["wrong_length"]}` |
| `"ATNC"` | Invalid character | `{"sequence": "ATNC", "valid": False, "errors": ["invalid_characters"]}` |
| `"ATN"` | Multiple errors | `{"sequence": "ATN", "valid": False, "errors": ["wrong_length", "invalid_characters"]}` |


Invalid reads should not crash your program.

### 💻 Enter your code here

````python
def validate_read(sequence, reference):
    """This funtion validates and normalises an NGS read agaisnt a reference sequence of choice.
    Action: input reads are converted into uppercase, it then checks the lengths of the read to see if they match the reference sequence.
    It also checks if the reads only contain the standard DNA bases (A, T, C, G).

    Arguments:
      sequence (str): The reads that should be validated
      reference (str): The reference sequence to check agaisnt

    What is returned:
      A dictionary containing (per read):
        - 'sequence' (str): uppercase version of the provided reads.
        - 'status' (str): 'valid' if it passes the tests, 'invalid' if not.
        - 'errors' (list of str): A list of the errors (if there are any, if not empty), the options are: 'wrong_length' and 'invalid_characters'.
    """
    summary_dict = {}
    sequence = sequence.upper()
    summary_dict['sequence'] = sequence
    #Go through every check and seperate into labels
    if len(sequence) == len(reference) and all(char in "ATGC" for char in sequence) :
      summary_dict['status'] = 'valid'
      summary_dict['errors'] = []
    elif len(sequence) == len(reference) :
      summary_dict['status'] = 'invalid'
      summary_dict['errors'] = ['invalid_characters']
    elif all(char in "ATGC" for char in sequence) :
      summary_dict['status'] = 'invalid'
      summary_dict['errors'] = ['wrong_length']
    else :
      summary_dict['status'] = 'invalid'
      summary_dict['errors'] = ['wrong_length', 'invalid_characters']
    return summary_dict
````

### 💻 Enter your code here

````python
def find_mutations(sequence, reference):
    """Indentifies the nucleotide mismatches between reference and valid read sequences.

    Arguments:
      sequence (str): Validated nucleotide read sequence containing only uppercase bases.
      reference (str): Reference nucleotide sequence of identical length containing
        only uppercase bases.

    Returns:
      A list of dictionaries, one per mismatch, ordered by position.
          Each dictionary contains:
             - 'position' (int): index of the mismatch in the sequence.
             - 'reference_base' (str): The expected nucleotide from the reference.
             - 'read_base' (str): The observed nucleotide from the read.
          Returns an empty list if no mismatches are detected.
    """
    mutation_list = []
    for index, char in enumerate(sequence) :
        if sequence[index] != reference[index] :
            mutation_dict = {
                'position': index + 1,
                'reference_base': reference[index],
                'read_base': sequence[index]
            }
            mutation_list.append(mutation_dict)
    return mutation_list
````

### 💻 Enter your code here

````python
def analyse_reads(reads, reference):
    """Integrates validate_read and find_mutations across all input reads.
  Each read dictionary has its uppercase sequence, validation status,
  error labels, and any detected mutations.

  Args:
      reads (list of dict): Each dictionary contains 'read_id',
          'sample', 'group', and 'sequence'.
      reference (str): Reference sequence (A, T, C, G) used
          for length checking and mismatch identification.

  Returns:
      list of dict: One per input read, each containing:
          - 'read_id' (str)
          - 'sample' (str)
          - 'group' (str)
          - 'sequence' (str): Uppercase read sequence.
          - 'status' (str): 'valid' or 'invalid'.
          - 'errors' (list of str): List of validation error labels, empty if valid.
          - 'mutations' (list of dict): Detected mismatch dictionaries for valid reads;
            empty list for invalid reads or perfectly matching reads.
    """
    analysed = []
    for index, num in enumerate(reads) :
      read_copy = reads[index].copy()
      validate = validate_read(read_copy['sequence'], reference)
      read_copy.update(validate)
      if read_copy.get('status') == 'valid' :
        mutations = find_mutations(read_copy['sequence'], reference)
        read_copy['mutations'] = mutations
      else :
        read_copy['mutations'] = []
      analysed.append(read_copy)
    return analysed
````

````python
def summarise_samples(analysed_reads):
    """Summarise mismatch frequencies, substitution info, and mutation locations by sample.

Calculates valid read percentages containing mismatches, overall mismatch
events, substitution frequencies, and identifies the top three mutation spots (if there is a tie-break, it picks the 3 earliest mutations).
Displays a formatted summary table and returns the nested dictionaries.

Args:
    analysed_reads (list of dict): List of read dictionaries produced by
        'analyse_reads', each containing 'sample', 'group', 'status', and 'mutations'.

Returns:
    dict: A dictionary mapping each sample name to a nested dictionary containing:
        - 'group' (str): 'treatment' or 'control'.
        - 'valid_reads' (int): Count of reads passing validation checks.
        - 'invalid_reads' (int): Count of reads failing validation checks.
        - 'mismatch_reads' (int): Count of valid reads containing at least one mutation.
        - 'mismatch_percentage' (float or None): Percentage of valid reads with mutations,
          or None if no valid reads exist.
        - 'mismatch_events' (int): Total count of individual nucleotide substitutions.
        - 'substitution_counts' (dict): Occurrences of each transition type (e.g. 'A>G').
        - 'position_counts' (dict): Total mutation count listed by sequence position.
        - 'top_positions' (list of int): Up to 3 positions with the highest mutation counts,
          in the case of a tie-break it picks the 3 earliest mutations. Returns an empty dictionary if the input is empty.
"""
    # 1. Group reads by sample
    grouped_by_sample = {}
    for read in analysed_reads:
        sample_name = read.get('sample')
        if sample_name:
            if sample_name not in grouped_by_sample:
                grouped_by_sample[sample_name] = []
            grouped_by_sample[sample_name].append(read)

    summary = {}

    # 2. Calculating the statistics for each sample
    for sample_name, reads_list in grouped_by_sample.items():
        sample_group = reads_list[0].get('group') if reads_list else None

        valid_count = 0
        invalid_count = 0
        mismatch_read_count = 0
        mismatch_events = 0
        substitution_counts = {}
        position_counts = {}

        for read in reads_list:
            if read.get('status') == 'valid':
                valid_count += 1
                mutations = read.get('mutations', [])

                if len(mutations) > 0:
                    mismatch_read_count += 1

                for event in mutations:
                    mismatch_events += 1

                    # Track substitution types (e.g. "A>T")
                    sub_key = f"{event['reference_base']}>{event['read_base']}"
                    substitution_counts[sub_key] = substitution_counts.get(sub_key, 0) + 1

                    # Track position counts
                    pos = event['position']
                    position_counts[pos] = position_counts.get(pos, 0) + 1
            else:
                invalid_count += 1

        # Calculate mismatch percentage
        if valid_count > 0:
            mismatch_percentage = (mismatch_read_count / valid_count) * 100.0
        else:
            mismatch_percentage = None

        # Determine top 3 positions (highest count descending; tie-break by earlier position)
        sorted_positions = sorted(
            position_counts.keys(),
            key=lambda p: (-position_counts[p], p)
        )
        top_positions = sorted_positions[:3]

        # Summary for this sample
        summary[sample_name] = {
            'group': sample_group,
            'valid_reads': valid_count,
            'invalid_reads': invalid_count,
            'mismatch_reads': mismatch_read_count,
            'mismatch_percentage': mismatch_percentage,
            'mismatch_events': mismatch_events,
            'substitution_counts': substitution_counts,
            'position_counts': position_counts,
            'top_positions': top_positions
        }

    # 3. Display results in a formatted text table using f-strings
    print(f"{'Sample':<12}{'Group':<12}{'Valid':>8}{'Invalid':>9}{'Mismatches':>12}{'Mismatch %':>12}{'Events':>8}{'Top Positions':>16}")
    print("-" * 89)
    for header, data in summary.items():
        percent_display = f"{data['mismatch_percentage']:>11.2f}%" if data['mismatch_percentage'] is not None else f"{'N/A':>12}"
        top_pos_display = ", ".join(map(str, data['top_positions']))
        print(f"{header:<12}{data['group']:<12}{data['valid_reads']:>8}{data['invalid_reads']:>9}{data['mismatch_reads']:>12}{percent_display}{data['mismatch_events']:>8}{top_pos_display:>16}")

    return summary


````

````python
#@title 📊 calculate_statistics function

from scipy import stats
import math
import warnings


def calculate_statistics(control_percentages, treated_percentages):
    """Compare independent biological samples using a two-sided Welch's t-test."""
    unavailable = {"statistic": None, "p_value": None}

    # Require at least two samples per group and finite numerical values.
    if len(control_percentages) < 2 or len(treated_percentages) < 2:
        return unavailable

    values = list(control_percentages) + list(treated_percentages)
    if any(value is None or not math.isfinite(value) for value in values):
        return unavailable

    # No within-group variation: the t-test cannot estimate uncertainty.
    if (
        len(set(control_percentages)) == 1
        and len(set(treated_percentages)) == 1
    ):
        return unavailable

    with warnings.catch_warnings():
        warnings.simplefilter("ignore", RuntimeWarning)
        result = stats.ttest_ind(
            treated_percentages,
            control_percentages,
            equal_var=False,
            alternative="two-sided"
        )

    if not (
        math.isfinite(result.statistic)
        and math.isfinite(result.pvalue)
    ):
        return unavailable

    return {
        "statistic": float(result.statistic),
        "p_value": float(result.pvalue)
    }
````
